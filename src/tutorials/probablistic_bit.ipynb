```python
import numpy as np
import sympy as sp
```

```python
p1 = np.array([[1/4],
               [3/4]])

sp.Matrix(p1)
```

```python
n_samples = 100
samples = np.random.choice([0, 1], size=n_samples, p=p1.reshape(-1))
# print(samples)
```

```python
zeros = np.count_nonzero(samples == 0)/n_samples
ones = np.count_nonzero(samples == 1)/n_samples

print(zeros, ones)
```

```python
p2 = np.array([[1/2],
               [1/2]])
```

```python
p = np.kron(p2, p1)
sp.Matrix(p)
```
