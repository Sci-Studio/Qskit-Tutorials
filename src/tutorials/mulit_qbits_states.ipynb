```python
import numpy as np
from qiskit.quantum_info import Statevector
```

```python
q = Statevector.from_label('+++')
q.draw('latex')
```

```python
q = np.sqrt(1/2) * (Statevector.from_label('00') + Statevector.from_label('11'))
q.draw('latex')
```

```python
q = 1/2 * (Statevector.from_label('00') + Statevector.from_label('01') + Statevector.from_label('10') - Statevector.from_label('11'))
q.draw('latex')
```

```python
q = 1/np.sqrt(2) * (Statevector.from_label('0+') + Statevector.from_label('1-'))
q.draw('latex')
```
