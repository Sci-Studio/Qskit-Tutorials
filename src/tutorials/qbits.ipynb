```python
import numpy as np
from qiskit.quantum_info import Statevector
```

```python
ket_0 = Statevector([1,0])
ket_0
```

```python
ket_0.probabilities()
```

```python
ket_p = Statevector([np.sqrt(1/2), np.sqrt(1/2)])
ket_p
```

```python
ket_p.probabilities()
```

```python
ket_p.draw('latex', convention='vector')
```
