```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
```

```python
qc = QuantumCircuit(3)
qc.z(2)
qc.h(1)
qc.x(0)
qc.draw()
```

```python
Operator(qc).draw('latex')
```

```python
X = Operator.from_label('X')
H = Operator.from_label('H')
Z = Operator.from_label('Z')
```

```python
ZHX = Operator(np.kron(Z, np.kron(H, X)))
ZHX.draw('latex')
```

```python
qc = QuantumCircuit(2)
qc.h(range(2))
qc.cz(1,0)
qc.draw()
```

```python
Operator(qc).draw('latex')
```

```python
from qiskit.quantum_info import Statevector
```

```python
Statevector(qc).draw('latex')
```

```python
I = Operator.from_label('I')
Z = Operator.from_label('Z')
pi0 = Operator.from_label('0')
pi1 = Operator.from_label('1')
```

```python
CZ = Operator(np.kron(pi0, I) + np.kron(pi1,Z))
CZ.draw('latex')
```

```python
qc = QuantumCircuit(2)
qc.cx(1,0,ctrl_state='0')
qc.draw()
```

```python
Operator(qc).draw('latex')
```

```python
from qiskit.circuit.library import PhaseGate, YGate
```

```python
mcp = PhaseGate(np.pi/2).control(3, ctrl_state='010')
mcy = YGate().control(3, ctrl_state='110')

qc = QuantumCircuit(4)
qc.append(mcp, [3,2,1,0])
qc.append(mcy, [3,2,1,0])
qc.draw()
```
