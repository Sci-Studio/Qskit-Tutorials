```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
ceta = np.pi/10
qc = QuantumCircuit(1)
qc.ry(ceta, 0)
qc.draw()
```

```python
Statevector(qc).draw('latex')
```

```python
qc_c = QuantumCircuit(2)
qc_c.ry(ceta, 1)
qc_c.cx(1,0)
qc_c.draw()
```

```python
c = Statevector(qc_c)
```

```python
qc_d = QuantumCircuit(2)
qc_d.ry(ceta, 1)
qc_d.ry(ceta, 0)
qc_d.draw()
```

```python
d = Statevector(qc_d)
```

```python
from qiskit.visualization import plot_histogram
```

```python
c_counts = c.sample_counts(1000, [0])
d_counts = d.sample_counts(1000, [0])
plot_histogram([c_counts, d_counts])
```
