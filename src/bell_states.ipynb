```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
qc = QuantumCircuit(2)
qc.x([1,0])
qc.barrier()
qc.h(1)
qc.cx(1,0)
qc.draw()
```

```python
phip = Statevector(qc)
phip.draw('latex')
```

```python
xip = Statevector(qc)
xip.draw('latex')
```

```python
phim = Statevector(qc)
phim.draw('latex')
```

```python
xim = Statevector(qc)
xim.draw('latex')
```
